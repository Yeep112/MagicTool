# SQUID（M–T、M–H）

和軟體用的是**同一份分析程式**（`magic/`），結果完全一樣；這裡適合自由探索、試新的分析。

- 第一格把 `DATA` 改成你的資料夾（沒填就用合成的範例資料）
- `run(機台, 分析名稱, [檔案], 參數=值…)` 回傳 `(fig, res)`：`res.text` 文字結果、`res.tables` 表格（pandas）
- 參數名稱和軟體參數面板一樣，可以用 `list_analyses("SQUID")` 看有哪些分析
- **上傳 GitHub 前會自動清掉輸出**（publish.bat），圖和數值不會被公開


```python
%matplotlib inline
import os, sys
sys.path.insert(0, os.path.abspath(".."))      # 沒有 pip install 也能 import magic（notebooks 的上一層就是工具資料夾）
from magic.runner import run, list_analyses
from magic.readers import load
from magic.core.plotstyle import apply
apply()                                          # 中文字型、配色和軟體一樣

DATA = r""   # ← SQUID 資料夾，例如 r"D:\Yeep\01 Magic Tool\SQUID\Data\20260918_In"
if not DATA:                                     # 沒填就用範例（合成資料）
    from magic.core.settings import data_dir
    from magic.testdata import write_all
    write_all(os.path.join(str(data_dir()), "demo"))
    DATA = os.path.join(str(data_dir()), "demo", "SQUID")
print("資料夾：", DATA)
print("可用的分析：", list_analyses("SQUID"))
sorted(os.listdir(DATA))
```

## M–T 與 Tc（ZFC / FC 自動分開）

```python
files = sorted(f for f in os.listdir(DATA) if f.lower().endswith(".dat"))
fp = os.path.join(DATA, [f for f in files if "MT" in f or "In" in f][0])   # ← 換成你的檔名
fig, res = run("SQUID", "M–T 與 Tc", [fp], method="50%（中點）")
print(res.text)
fig
```

## 曲線檢視（單位自動換成 μemu…）

```python
fig, res = run("SQUID", "曲線檢視", [fp], x="Temperature (K)", y="Moment (emu)", x_range="0, 10")
fig
```

## M–H 迴線

```python
mh = [f for f in files if "MH" in f]
if mh:
    fig, res = run("SQUID", "M–H 迴線", [os.path.join(DATA, mh[0])], bg_field=None)
    print(res.text)
    display(fig)
```
